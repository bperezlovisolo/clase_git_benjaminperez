```python
usa_counts = proyect_sharks.loc[proyect_sharks['Country'] == 'USA']
usa_counts
```

```python
usa_counts["Year"].value_counts()
```

```python
#Trabajaremos con los datos registrados en Estados Unidos, a partir del año
usa_counts1 = usa_counts[usa_counts['Year'].between(2000,2026)]
usa_counts1
```

```python
#Se muestran los estados donde mayor episodios de ataques de tiburones hubo, en formato descendiente.  
usa_counts1["State"] = usa_counts1['State'].str.strip()
print (usa_counts1['State'].value_counts(normalize=True).head(5) * 100)
```

```python
correcciones = {
    'M': 'M',
    'F': 'F',
    'M ': 'M', # Por si acaso
    'lli': 'M', # Suponiendo que 'lli' era una 'M' mal escrita
    'm': 'M'
}
usa_counts1['Sex'] = usa_counts1['Sex'].map(correcciones)

usa_counts1["Sex"] = usa_counts1['Sex'].str.strip()
print (usa_counts1['Sex'].value_counts(normalize=True) * 100)
```

```python
usa_counts1['Age'] = pd.to_numeric(usa_counts1['Age'], errors='coerce')
bins = [0, 12, 18, 35, 60, 100]
labels = ['Child', 'Teen', 'Young Adult', 'Adult', 'Senior']

usa_counts1['Age_Group'] = pd.cut(usa_counts1['Age'], bins=bins, labels=labels)
print(usa_counts1['Age_Group'].value_counts(normalize=True) * 100)
```

```python
top_5 = usa_counts1['Activity'].value_counts().head(5).index
usa_counts1['Activity'] = usa_counts1['Activity'].where(usa_counts1['Activity'].isin(top_5), 'Other')
print(usa_counts1['Activity'].value_counts(normalize=True) * 100)

```
